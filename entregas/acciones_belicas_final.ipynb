- **Variables categóricas/texto (str):** información geográfica (municipio, departamento, región, código DANE), información sobre los actores (grupo armado 1, 2 y 3 con sus descripciones), modalidad del evento (combate, hostigamiento, emboscada, bombardeo, etc.), iniciativa, tipo de unidad atacada y ventaja militar.
- **Variables geoespaciales (str):** coordenadas de latitud-longitud del evento.

#### **1.4. Integración y transformaciones iniciales**

Se realizan las transformaciones mínimas necesarias para poder explorar el dataset: conversión de tipos de datos, tratamiento de la columna de coordenadas y creación de variables derivadas útiles para el análisis posterior.

```python
# ── 1. Renombrar columnas de muertes para claridad semántica ─────────────────
RENAME_MUERTES = {
    'Militares':                'Muertes Militares',
    'Policías':                 'Muertes Policías',
    'Otras Fuerzas Armadas Estatales': 'Muertes Otras Fuerzas Armadas Estatales',
    'Agentes del Estado Sin Información': 'Muertes Agentes del Estado Sin Información',
    'Total Agentes del Estado': 'Muertes Total Agentes del Estado',
    'Guerrilleros':             'Muertes Guerrilleros',
    'Paramilitares':            'Muertes Paramilitares',
    'Grupos Posdesmovilización':'Muertes Grupos Posdesmovilización',
    'Combatientes Sin Información': 'Muertes Combatientes Sin Información',
    'Otros Grupos Armados Organizados': 'Muertes Otros Grupos Armados Organizados',
    'Total Combatientes de Grupos Armados Organizados': 'Muertes Total Combatientes GAO',
    'Total Combatientes':       'Muertes Total Combatientes',
    'Personas Sin Información': 'Muertes Personas Sin Información',
    'Total Civiles':            'Muertes Total Civiles',
}
df.rename(columns=RENAME_MUERTES, inplace=True)
print(f"Columnas renombradas ({len(RENAME_MUERTES)}):")
for old, new in RENAME_MUERTES.items():
    print(f"  {old:<55} → {new}")

# ── 2. Tipos string ─────────────────────────────────────────────────────────
str_cols = ['ID Caso', 'ID Caso Relacionado', 'Código DANE de Municipio',
            'Municipio', 'Departamento', 'Región', 'latitud-longitud']
for col in str_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).replace('nan', pd.NA)

# ── 3. Tipos entero (Int64 nullable) ────────────────────────────────────────
int_cols = [
    'Año', 'Mes', 'Día',
    'Lesionados Civiles', 'Capturados', 'Lesionados Combatientes',
    'Total de Víctimas del Caso',
] + list(RENAME_MUERTES.values())

for col in int_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')

# ── 4. Tipos categóricos ────────────────────────────────────────────────────
cat_cols = [
    'Modalidad', 'Iniciativa', 'Tipo de Unidad Atacada',
    'Grupo Armado 1', 'Descripción Grupo Armado 1',
    'Grupo Armado 2', 'Descripción Grupo Armado 2',
    'Grupo Armado 3', 'Descripción Grupo Armado 3',
    'Grupo al que Pertenecen los Capturados', 'Ventaja Militar',
]
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')

# ── 5. Extracción de coordenadas ────────────────────────────────────────────
def parse_point(val):
    if pd.isna(val):
        return pd.Series([np.nan, np.nan])
    try:
        coords = str(val).replace('POINT (', '').replace(')', '').strip().split()
        lon, lat = float(coords[0]), float(coords[1])
        return pd.Series([lat, lon])
    except Exception:
        return pd.Series([np.nan, np.nan])

if 'latitud-longitud' in df.columns:
    df[['Latitud', 'Longitud']] = df['latitud-longitud'].apply(parse_point)

# ── 6. Variables derivadas ──────────────────────────────────────────────────
df['Año_int'] = df['Año'].fillna(0).astype(int)

bins_decada = list(range(1950, 2031, 10))
labels_decada = [f"{d}s" for d in bins_decada[:-1]]
df['Década'] = pd.cut(df['Año_int'], bins=bins_decada, labels=labels_decada, right=False)

df['Tiene_Víctimas'] = (df['Total de Víctimas del Caso'] > 0).astype(int)

# ── Resumen ─────────────────────────────────────────────────────────────────
print(f"\nTransformaciones completadas:")
print(f"  - Columnas renombradas (muertes):   {len(RENAME_MUERTES)}")
print(f"  - Columnas → string:                {len(str_cols)}")
print(f"  - Columnas → Int64:                 {len(int_cols)}")
print(f"  - Columnas → category:              {len(cat_cols)}")
print(f"  - Coordenadas extraídas:            Latitud, Longitud")
print(f"  - Variables derivadas:              Año_int, Década, Tiene_Víctimas")
print(f"  - Dimensiones finales:              {df.shape[0]:,} filas × {df.shape[1]} columnas")

print("\nVerificación de tipos de datos:")
print(df.dtypes.to_string())
```

#### Hitos del conflicto armado colombiano (contexto histórico)

##### 1964: Fundación de las FARC y comienzo del conflicto moderno

El surgimiento de las guerrillas contemporáneas a mediados de la década de 1960 marca el inicio del conflicto armado colombiano tal como se conoce. En mayo de 1964, el gobierno lanzó la **Operación Marquetalia** contra un enclave campesino insurgente en el sur de Tolima; de esa resistencia nacerían las **Fuerzas Armadas Revolucionarias de Colombia (FARC)**. Este evento se considera el mito fundacional de las FARC-EP. La fundación de las FARC (junto con el ELN, surgido también en 1964) es un hito porque marca el inicio organizado del enfrentamiento armado entre el Estado y guerrillas de inspiración revolucionaria, dando comienzo a un conflicto prolongado. *(Comisión para el Esclarecimiento de la Verdad, la Convivencia y la No Repetición, s. f.).*

##### 1984: Acuerdos de La Uribe – primer intento de paz

El gobierno de **Belisario Betancur (1982–1986)** protagonizó el primer proceso de paz serio con las guerrillas. El **28 de marzo de 1984** se firmaron los **Acuerdos de La Uribe** entre el gobierno y las FARC-EP, estableciendo un cese al fuego bilateral y una agenda de diálogos en plena guerra. Por primera vez el Estado reconoció a una guerrilla como interlocutor político válido y abrió una vía negociada para transformar la confrontación. Los acuerdos incluían tregua, amnistía para delitos políticos y la posibilidad de incorporación a la vida legal mediante la creación de un partido político —la **Unión Patriótica (UP)**— como resultado de la apertura democrática pactada. Aunque este proceso fracasó pocos años después, los Acuerdos de La Uribe son considerados un hito histórico al representar el inicio de la búsqueda de una salida negociada en medio de la guerra. *(Centro Nacional de Memoria Histórica [CNMH], 2025).*

##### 1991: Nueva Constitución Política y desmovilización de guerrillas

La promulgación de la **Constitución de 1991** significó un hito trascendental en el plano político-institucional, con efectos directos sobre el conflicto. En plena guerra, Colombia logró un “nuevo contrato social”, superando el orden excluyente de la carta de 1886. Este cambio fue posible gracias a negociaciones de paz con varias guerrillas en 1989–1990, que llevaron a dejación de armas y participación política. En particular, el **M-19** se desmovilizó y se convirtió en partido político (**Alianza Democrática M-19**), contribuyendo activamente a la Asamblea Constituyente. Asimismo, en 1991 se desmovilizaron el **EPL**, el **PRT** y el **Movimiento Armado Quintín Lame (MAQL)**, integrándose muchos excombatientes a la vida civil y política. *(CIDOB, 2015; Comisión para el Esclarecimiento de la Verdad, la Convivencia y la No Repetición, s. f.).*

##### 2002: Ruptura del proceso del Caguán y pico de violencia del conflicto

El año 2002 marcó un hito negativo pero decisivo: la fractura de las negociaciones de paz del **Caguán** y el inicio de la etapa más violenta del conflicto armado. Entre 1999 y 2002, el gobierno de **Andrés Pastrana** estableció una zona desmilitarizada en **San Vicente del Caguán** para dialogar con las FARC, pero el proceso fracasó. El punto de quiebre ocurrió el **20 de febrero de 2002**, cuando la guerrilla secuestró a un senador durante los diálogos; tras ello, el presidente Pastrana dio por terminado el proceso, anuló la zona de distensión y rompió cualquier reconocimiento político a las FARC. Al día siguiente (**21 de febrero**), el gobierno lanzó una operación para que las Fuerzas Militares retomaran el control de la zona despejada. Esta ruptura significó un giro dramático: se pasó de buscar la paz negociada a una ofensiva militar de gran escala bajo el gobierno de **Álvaro Uribe (2002–2010)**. Como consecuencia, el período de 2002 coincide con el pico máximo de intensidad del conflicto: nunca antes hubo tantos hechos de violencia como a comienzos de la década de 2000. Además, entre 1998 y 2002 ocurrió el mayor número de masacres conocidas en Colombia (1.620 en esos años), perpetradas en gran medida por grupos paramilitares unificados bajo las **AUC** desde 1997. La expansión del paramilitarismo en la segunda mitad de los 90 y la ofensiva guerrillera condujeron a niveles récord de muertes y violaciones de derechos humanos alrededor de 2002. *(Comisión Interamericana de Derechos Humanos, 2004; Comisión para el Esclarecimiento de la Verdad, la Convivencia y la No Repetición, s. f.).*

##### 2016: Acuerdo final de paz Gobierno–FARC

El **Acuerdo Final de Paz de 2016** entre el gobierno colombiano y las FARC-EP constituye un hito histórico al terminar formalmente el enfrentamiento armado de más de medio siglo con la guerrilla más grande del país. Tras un proceso de negociaciones de cuatro años (2012–2016) en **La Habana**, el **24 de agosto de 2016** se alcanzó la firma de un acuerdo completo. Este hecho cerró un capítulo central del conflicto: las FARC dejaron las armas y se convirtieron en partido político en 2017, reduciendo significativamente la violencia en muchas regiones. Si bien persistieron otros actores armados (ELN, disidencias, EPL, bandas criminales), el acuerdo de 2016 representa el punto de inflexión más importante hacia la paz en Colombia en décadas recientes. *(CIDOB, 2015).*

```python
print("=" * 80)
print("4. ANÁLISIS DE VALORES ATÍPICOS (OUTLIERS)")
print("=" * 80)

outlier_cols = ['Total de Víctimas del Caso', 'Muertes Total Civiles', 'Muertes Total Combatientes',
                'Muertes Total Agentes del Estado', 'Muertes Total Combatientes GAO']

outlier_summary = []
for col in outlier_cols:
    if col in df.columns:
        data = df[col].dropna()
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        n_outliers = ((data < lower) | (data > upper)).sum()
        outlier_summary.append({
            'Variable': col,
            'Q1': Q1,
            'Q3': Q3,
            'IQR': IQR,
            'Límite inferior': lower,
            'Límite superior': upper,
            'N° outliers': n_outliers,
            '% outliers': round(n_outliers / len(data) * 100, 2),
        })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

Se clasifican todas las variables del dataset, se identifica si existe una variable objetivo y se analiza la distribución de las categorías principales para detectar posibles desbalanceos.

```python
print("=" * 80)
print("CLASIFICACIÓN DE VARIABLES DEL DATASET")
print("=" * 80)

var_classification = []
for col in df.columns:
    dtype = str(df[col].dtype)
    n_unique = df[col].nunique()

    if col in ['ID Caso', 'ID Caso Relacionado', 'Código DANE de Municipio', 'latitud-longitud']:
        role = 'Identificador'
    elif col in ['Año', 'Mes', 'Día', 'Año_int', 'Década']:
        role = 'Temporal'
    elif col in ['Departamento', 'Municipio', 'Región', 'Latitud', 'Longitud']:
        role = 'Geográfica'
    elif col in ['Modalidad', 'Iniciativa', 'Tipo de Unidad Atacada', 'Ventaja Militar']:
        role = 'Caracterización del evento'
    elif 'Grupo Armado' in col or 'Descripción Grupo' in col:
        role = 'Actores del conflicto'
    elif col == 'Tiene_Víctimas':
        role = 'Variable derivada (binaria)'
    elif 'Muertes' in col:
        role = 'Muertes en combate'
    else:
        role = 'Víctimas / Afectación'

    if df[col].dtype in ['float64', 'int64', 'int32']:
        if n_unique <= 2:
            var_type = 'Binaria'
        elif n_unique <= 20:
            var_type = 'Numérica discreta'
        else:
            var_type = 'Numérica continua'
    elif df[col].dtype == 'object':
        if n_unique <= 10:
            var_type = 'Categórica nominal (baja cardinalidad)'
        elif n_unique <= 50:
            var_type = 'Categórica nominal (media cardinalidad)'
        else:
            var_type = 'Categórica nominal (alta cardinalidad)'
    elif df[col].dtype.name == 'category':
        var_type = 'Categórica ordinal'
    else:
        var_type = str(df[col].dtype)

    var_classification.append({
        'Variable': col,
        'Tipo Pandas': dtype,
        'Tipo Estadístico': var_type,
        'Rol en el análisis': role,
        'Valores únicos': n_unique,
    })

var_df = pd.DataFrame(var_classification)
var_df.index = range(1, len(var_df) + 1)
var_df
```

##### **Análisis de la variable `Tiene_Víctimas` como posible variable objetivo de clasificación**

La variable derivada `Tiene_Víctimas` es una variable binaria (0/1) que indica si una acción bélica produjo al menos una víctima. Podría utilizarse como variable objetivo en un modelo de **clasificación binaria** para predecir si un evento genera víctimas a partir de las características del contexto. Las columnas de muertes (Muertes Militares, Muertes Guerrilleros, etc.), lesionados y capturados **deben excluirse** como predictores, ya que `Tiene_Víctimas` se deriva directamente de `Total de Víctimas del Caso`, que a su vez es la suma de dichas columnas.

```python
print("=" * 80)
print("DISTRIBUCIÓN DE LA VARIABLE OBJETIVO: Tiene_Víctimas")
print("=" * 80)

victimas_dist = df['Tiene_Víctimas'].value_counts().sort_index()
labels_tv = {0: 'Sin víctimas (0)', 1: 'Con víctimas (1)'}

total = len(df)
for val, count in victimas_dist.items():
    print(f"  {labels_tv[val]:<25} {count:>10,}  ({count/total*100:.1f}%)")

ratio = victimas_dist.max() / victimas_dist.min()
print(f"\n  Razón de desbalanceo (clase mayoritaria / minoritaria): {ratio:.2f}x")

print("\n" + "=" * 80)
print("VARIABLES EXCLUIDAS (data leakage)")
print("=" * 80)
print("  Tiene_Víctimas se deriva de Total de Víctimas del Caso,")
print("  que a su vez es la suma de las columnas de muertes,")
print("  lesionados y capturados. Estas variables deben excluirse")
print("  de cualquier modelo predictivo:")

leak_cols = list(RENAME_MUERTES.values()) + [
    'Lesionados Civiles', 'Lesionados Combatientes', 'Capturados',
    'Muertes Total Civiles', 'Muertes Total Combatientes',
    'Total de Víctimas del Caso', 'Tiene_Víctimas',
]
for c in sorted(set(leak_cols)):
    if c in df.columns:
        print(f"    ✗ {c}")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

bars = axes[0].bar([labels_tv[0], labels_tv[1]],
                   [victimas_dist.get(0, 0), victimas_dist.get(1, 0)],
                   color=[COLORS[0], COLORS[2]], edgecolor='black')
for bar in bars:
    h = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2., h + 150,
                 f'{int(h):,}', ha='center', fontweight='bold')
axes[0].set_title('Distribución de Tiene_Víctimas', fontweight='bold')
axes[0].set_ylabel('Número de registros')

feature_cats = ['Modalidad', 'Grupo Armado 1', 'Iniciativa', 'Ventaja Militar']
rates = {}
for col in feature_cats:
    grp = df.groupby(col)['Tiene_Víctimas'].mean().sort_values(ascending=False)
    rates[col] = grp

col_pick = 'Modalidad'
r = rates[col_pick]
r.plot(kind='barh', ax=axes[1], color=COLORS[4], edgecolor='black')
axes[1].set_title(f'Tasa de víctimas por {col_pick}', fontweight='bold')
axes[1].set_xlabel('Proporción de eventos con víctimas')
axes[1].invert_yaxis()
for i, v in enumerate(r.values):
    axes[1].text(v + 0.005, i, f'{v:.2%}', va='center', fontsize=9)

plt.suptitle('Variable objetivo candidata para clasificación: Tiene_Víctimas',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
save_fig(fig, '09_tiene_victimas_clasificacion')
plt.show()
```

#### Hallazgos principales

1. **Estructura del dataset:** El conjunto de datos contiene 38,333 registros de acciones bélicas con 39 variables originales que cubren dimensiones temporales, geográficas, de actores y de víctimas. Se renombraron 14 columnas de conteo de muertes para incluir el prefijo "Muertes", clarificando que hacen referencia a personas fallecidas en combate. ### **1.1. Valores faltantes**
---

En la [Fase 2 (sección 2.3 — Análisis de Valores Faltantes)](#fase2) se identificaron los siguientes patrones de valores faltantes:

| Columna | % Nulos | Observación |
|---------|---------|-------------|
| Grupo Armado 3 | 97.3% | La gran mayoría de acciones bélicas involucran solo 2 actores |
| Descripción Grupo Armado 3 | 97.4% | Asociada directamente a Grupo Armado 3 |
| ID Caso Relacionado | 89.1% | Campo de referencia cruzada, mayormente vacío |
| Grupo al que Pertenecen los Capturados | 93.1% | La captura no es resultado frecuente de acciones bélicas |
| Tipo de Unidad Atacada | 0.4% técnico + 58.9% "SIN INFORMACIÓN" | Categórica con alta proporción de "SIN INFORMACIÓN" |
| Región | 1.6% | Valores nulos reales más valores "SIN INFORMACIÓN" |

Adicionalmente, se detectó el uso sistemático del valor **"SIN INFORMACION"** como categoría explícita para indicar ausencia de información en las variables categóricas. Este tratamiento heterogéneo (nulos reales vs. cadenas "SIN INFORMACION") dificulta el análisis posterior, por lo que se unificará.

**Decisiones técnicas:**

1. **Unificación de valores "SIN INFORMACION":** Se reemplazarán por `NaN` en las columnas geográficas y categóricas para tratamiento consistente. Esto permite que las funciones estándar de Pandas reconozcan estos valores como faltantes.
2. **Variables temporales con valor 0:** Los registros con `Año = 0`, `Mes = 0` o `Día = 0` representan eventos cuya fecha no pudo determinarse. Se marcarán como `NaN` para que no contaminen análisis temporales, pero se conservarán los registros ya que contienen información válida en otras dimensiones (geográfica, actores, víctimas).
3. **Columnas con >90% nulos:** Se abordarán en la sección de Selección de Características (sección 2), ya que su alta nulidad las convierte en candidatas para eliminación.

```python
print("=" * 80)
print("1.1. TRATAMIENTO DE VALORES FALTANTES")
print("=" * 80)

df_prep = df.copy()

# ── 1. Unificar "SIN INFORMACION" → NaN en columnas categóricas y de texto ──
sin_info_values = ['SIN INFORMACION', 'SIN INFORMACIÓN', 'NO APLICA', 'N/A', 'NA']
cols_sin_info = ['Departamento', 'Municipio', 'Región', 'Modalidad',
                 'Grupo Armado 1', 'Descripción Grupo Armado 1',
                 'Grupo Armado 2', 'Descripción Grupo Armado 2',
                 'Grupo Armado 3', 'Descripción Grupo Armado 3',
                 'Iniciativa', 'Ventaja Militar', 'Tipo de Unidad Atacada',
                 'Grupo al que Pertenecen los Capturados']

n_replaced_total = 0
print("\n--- Reemplazos 'SIN INFORMACIÓN' → NaN ---")
for col in cols_sin_info:
    if col in df_prep.columns:
        mask = df_prep[col].isin(sin_info_values)
        n_replaced = mask.sum()
        if n_replaced > 0:
            df_prep.loc[mask, col] = pd.NA
            print(f"  {col:<45} → {n_replaced:>6,} valores reemplazados")
            n_replaced_total += n_replaced

print(f"\n  Total de valores 'SIN INFORMACIÓN' unificados a NaN: {n_replaced_total:,}")

# ── 2. Variables temporales: valor 0 → NaN ───────────────────────────────────
print("\n--- Tratamiento de variables temporales con valor 0 ---")
for col in ['Año', 'Mes', 'Día']:
    if col in df_prep.columns:
        mask_zero = df_prep[col] == 0
        n_zero = mask_zero.sum()
        if n_zero > 0:
            df_prep.loc[mask_zero, col] = pd.NA
            print(f"  {col}: {n_zero:,} registros con valor 0 → NaN")

if 'Año_int' in df_prep.columns:
    df_prep['Año_int'] = df_prep['Año'].fillna(0).astype(int)

# ── 3. Resumen post-tratamiento ──────────────────────────────────────────────
print("\n--- Completitud tras tratamiento de faltantes ---")
null_after = df_prep.isna().sum()
total_cells = df_prep.shape[0] * df_prep.shape[1]
total_nulls_after = null_after.sum()
print(f"  Completitud general: {(1 - total_nulls_after / total_cells) * 100:.1f}%")
print(f"  (Nota: la completitud disminuye porque los 'SIN INFORMACIÓN' ahora se contabilizan como nulos reales)")
```

#### **Imputación de Región a partir de Municipio**

Dado que la variable `Región` presenta valores faltantes tras la unificación de "SIN INFORMACIÓN", se aprovecha la relación existente en el dataset entre `Municipio` y `Región`: si un municipio aparece en otros registros con su región asignada, se puede homologar esa región para los registros donde falta. Este procedimiento es válido porque la relación municipio-región es una propiedad geográfica fija (un municipio siempre pertenece a la misma región).

**Justificación:** Recuperar información de región sin recurrir a fuentes externas, utilizando la consistencia interna del propio dataset, mejora la completitud sin introducir sesgo ni suposiciones externas.

```python
print("=" * 80)
print("IMPUTACIÓN DE REGIÓN A PARTIR DE MUNICIPIO")
print("=" * 80)

n_region_missing_before = df_prep['Región'].isna().sum()
print(f"\nRegistros sin región antes de imputación: {n_region_missing_before:,}")

mun_region_map = (
    df_prep[df_prep['Región'].notna()]
    .groupby('Municipio')['Región']
    .agg(lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else pd.NA)
)

mask_region_missing = df_prep['Región'].isna() & df_prep['Municipio'].notna()
records_to_fill = df_prep.loc[mask_region_missing].copy()

n_filled = 0
n_not_found = 0
not_found_municipios = []

for idx, row in records_to_fill.iterrows():
    mun = row['Municipio']
    if mun in mun_region_map.index and pd.notna(mun_region_map[mun]):
        df_prep.loc[idx, 'Región'] = mun_region_map[mun]
        n_filled += 1
    else:
        n_not_found += 1
        if mun not in not_found_municipios:
            not_found_municipios.append(mun)

n_region_missing_after = df_prep['Región'].isna().sum()

print(f"\nResultados de la imputación:")
print(f"  Regiones imputadas exitosamente:       {n_filled:,}")
print(f"  Registros sin municipio conocido:       {n_not_found:,}")
print(f"  Registros sin región después:           {n_region_missing_after:,}")
print(f"  Reducción de nulos en Región:           {n_region_missing_before - n_region_missing_after:,}")

if not_found_municipios:
    print(f"\n  Municipios sin correspondencia de región ({len(not_found_municipios)}):")
    for m in not_found_municipios:
        n_records = df_prep[(df_prep['Municipio'] == m) & df_prep['Región'].isna()].shape[0]
        print(f"    • {m}: {n_records} registro(s)")

print(f"\n  Los registros restantes sin región corresponden a eventos donde")
print(f"  tanto la región como el municipio son desconocidos (NaN).")
```

### **1.5. Validación de coordenadas geográficas**
---

En la [Fase 2 (sección 2.3 — Validación de Coordenadas)](#fase2) se identificaron **5 registros** con coordenadas fuera del rango geográfico esperado para Colombia (Latitud: -5° a 13°, Longitud: -82° a -67°). Dado que se trata de un número reducido de registros, se realiza a continuación una inspección detallada de cada caso para determinar si se trata de errores de registro o de eventos que efectivamente ocurrieron fuera del territorio colombiano en el marco del conflicto armado.

```python
print("=" * 80)
print("1.5. VALIDACIÓN DE COORDENADAS GEOGRÁFICAS")
print("=" * 80)

lat_min, lat_max = -5, 13
lon_min, lon_max = -82, -67

mask_outside = ~(
    (df_prep['Latitud'].between(lat_min, lat_max)) &
    (df_prep['Longitud'].between(lon_min, lon_max))
) & df_prep['Latitud'].notna()

coords_outside = df_prep[mask_outside].copy()
print(f"\nRegistros con coordenadas fuera del territorio colombiano: {len(coords_outside)}")

cols_display = ['ID Caso', 'Año', 'Mes', 'Día', 'Municipio', 'Departamento', 'Región',
                'Modalidad', 'Grupo Armado 1', 'Descripción Grupo Armado 1',
                'Grupo Armado 2', 'Descripción Grupo Armado 2',
                'Latitud', 'Longitud', 'latitud-longitud',
                'Iniciativa', 'Ventaja Militar',
                'Total de Víctimas del Caso']

existing_cols = [c for c in cols_display if c in df_prep.columns]

print("\n" + "─" * 80)
for idx, (i, row) in enumerate(coords_outside.iterrows()):
    print(f"\n{'═' * 80}")
    print(f"REGISTRO {idx + 1} de {len(coords_outside)} — Índice: {i}")
    print(f"{'═' * 80}")
    for col in existing_cols:
        val = row[col]
        print(f"  {col:<45} → {val}")
    print(f"\n  📍 Coordenadas: ({row['Latitud']:.6f}, {row['Longitud']:.6f})")

    lat, lon = row['Latitud'], row['Longitud']
    if -4.2 <= lat <= 12.5 and -79.0 <= lon <= -66.8:
        print("  ⚠ Coordenada en zona limítrofe de Colombia")
    elif lat < -5 or lat > 13:
        print("  ❌ Latitud fuera del rango esperado para Colombia")
    if lon < -82 or lon > -67:
        print("  ❌ Longitud fuera del rango esperado para Colombia")
```

```python
print("=" * 80)
print("ANÁLISIS CONTEXTUAL DE REGISTROS CON COORDENADAS FUERA DE COLOMBIA")
print("=" * 80)

print("""
Para cada registro se examina la coherencia entre las coordenadas y los campos
de municipio, departamento y región, con el fin de determinar si:
  (a) El evento efectivamente ocurrió fuera de Colombia en el marco del conflicto
  (b) Las coordenadas son un error de registro

Criterios de decisión:
  - Si el municipio y departamento corresponden a localidades colombianas reales
    pero las coordenadas apuntan fuera → probable error de coordenadas
  - Si el municipio/departamento es "SIN INFORMACIÓN" y las coordenadas apuntan
    a zonas fronterizas o países vecinos → posible evento transfronterizo legítimo
  - Si las columnas del registro muestran indicios de actividad fuera del país
    (zona fronteriza, operaciones en territorio extranjero) → evento legítimo
""")

for idx, (i, row) in enumerate(coords_outside.iterrows()):
    lat, lon = row['Latitud'], row['Longitud']
    mun = row.get('Municipio', 'N/A')
    dep = row.get('Departamento', 'N/A')
    region = row.get('Región', 'N/A')
    modalidad = row.get('Modalidad', 'N/A')
    ga1 = row.get('Descripción Grupo Armado 1', 'N/A')
    ga2 = row.get('Descripción Grupo Armado 2', 'N/A')

    print(f"\n{'─' * 80}")
    print(f"Registro {idx + 1}: ID Caso = {row['ID Caso']}")
    print(f"  Coordenadas:  ({lat:.6f}, {lon:.6f})")
    print(f"  Municipio:    {mun}")
    print(f"  Departamento: {dep}")
    print(f"  Región:       {region}")
    print(f"  Modalidad:    {modalidad}")
    print(f"  Grupo 1:      {ga1}")
    print(f"  Grupo 2:      {ga2}")

    is_sin_info_mun = pd.isna(mun) or str(mun).strip() in ['', 'SIN INFORMACION', 'SIN INFORMACIÓN']
    is_sin_info_dep = pd.isna(dep) or str(dep).strip() in ['', 'SIN INFORMACION', 'SIN INFORMACIÓN']

    if is_sin_info_mun and is_sin_info_dep:
        print("  → Municipio y departamento sin información. Las coordenadas fuera del territorio")
        print("    pueden indicar un evento transfronterizo o un error de geolocalización.")
        print("    CLASIFICACIÓN: Posible evento sin localización precisa.")
    elif not is_sin_info_mun and not is_sin_info_dep:
        print(f"  → El registro indica municipio '{mun}' en '{dep}', que son localidades colombianas.")
        print(f"    Sin embargo, las coordenadas ({lat:.4f}, {lon:.4f}) no corresponden a Colombia.")
        print("    CLASIFICACIÓN: Probable error de coordenadas — se conserva el registro con coordenadas marcadas como inválidas.")
    else:
        print("  → Información parcial disponible. Requiere revisión caso a caso.")
        print("    CLASIFICACIÓN: Indeterminado — se conserva con marca.")
```

```python
print("=" * 80)
print("DECISIÓN SOBRE COORDENADAS FUERA DE RANGO")
print("=" * 80)

print("""
Dado el análisis anterior, se toma la siguiente decisión para los registros con
coordenadas fuera del territorio colombiano:

  1. Se crea una columna indicadora 'Coords_Validas' (True/False) que marca si
     las coordenadas caen dentro del rango geográfico de Colombia.
  2. Los registros NO se eliminan, ya que contienen información válida en las
     demás variables (modalidad, actores, víctimas, etc.).
  3. Para análisis geográficos que dependan de coordenadas (mapas, clusters
     espaciales), se filtrarán usando 'Coords_Validas == True'.
  4. Para los registros con municipio/departamento colombiano conocido pero
     coordenadas fuera de rango, se invalidan las coordenadas (NaN) conservando
     la información geográfica textual.

Esta decisión es conservadora: se preserva la máxima cantidad de información
sin contaminar los análisis espaciales con coordenadas incorrectas.
""")

df_prep['Coords_Validas'] = (
    df_prep['Latitud'].between(lat_min, lat_max) &
    df_prep['Longitud'].between(lon_min, lon_max)
) | df_prep['Latitud'].isna()

n_invalid_coords = (~df_prep['Coords_Validas']).sum()

for i, row in coords_outside.iterrows():
    mun = row.get('Municipio', '')
    dep = row.get('Departamento', '')
    is_sin_info = pd.isna(mun) or str(mun).strip() in ['', 'SIN INFORMACION', 'SIN INFORMACIÓN']
    if not is_sin_info:
        df_prep.loc[i, 'Latitud'] = np.nan
        df_prep.loc[i, 'Longitud'] = np.nan

print(f"Coordenadas marcadas como inválidas: {n_invalid_coords}")
print(f"Registros con coordenadas válidas:   {df_prep['Coords_Validas'].sum():,}")
print(f"Nueva columna agregada: 'Coords_Validas'")
```

### **3.1. Normalización de categorías textuales**
---

Las variables categóricas de texto (Municipio, Departamento, Región) pueden presentar variaciones en mayúsculas/minúsculas, espacios adicionales o caracteres especiales. Se estandarizan para garantizar consistencia en agrupaciones y cruces.

**Justificación:** Sin esta normalización, un mismo municipio podría aparecer como categorías distintas (por ejemplo, "MEDELLÍN" vs "Medellín" o con tildes), generando conteos incorrectos y agrupaciones fragmentadas en los análisis.

```python
print("=" * 80)
print("3.1. NORMALIZACIÓN DE CATEGORÍAS TEXTUALES")
print("=" * 80)

text_cols_normalize = ['Municipio', 'Departamento', 'Región']

for col in text_cols_normalize:
    if col in df_prep.columns:
        before_unique = df_prep[col].nunique()
        df_prep[col] = df_prep[col].astype(str).str.strip().str.upper()
        df_prep.loc[df_prep[col].isin(['NAN', 'NONE', '<NA>', '']), col] = pd.NA
        after_unique = df_prep[col].nunique()
        diff = before_unique - after_unique
        if diff > 0:
            print(f"  {col}: {before_unique} → {after_unique} categorías únicas (−{diff} por normalización)")
        else:
            print(f"  {col}: {after_unique} categorías únicas (sin cambios por normalización)")

cat_cols_normalize = ['Modalidad', 'Iniciativa', 'Ventaja Militar',
                      'Grupo Armado 1', 'Grupo Armado 2',
                      'Descripción Grupo Armado 1', 'Descripción Grupo Armado 2',
                      'Tipo de Unidad Atacada', 'Grupo al que Pertenecen los Capturados']

print("\n--- Verificación de categorías en variables categóricas ---")
for col in cat_cols_normalize:
    if col in df_prep.columns:
        n_cats = df_prep[col].nunique()
        print(f"  {col:<50} → {n_cats:>4} categorías")
```

### **3.2. Creación de variable Fecha**
---

Cuando los tres componentes temporales (`Año_int`, `Mes`, `Día`) están disponibles y son válidos, se construye una columna `Fecha` de tipo `datetime`. Esto facilita análisis temporales de resolución diaria, cálculo de intervalos entre eventos y series de tiempo.

**Justificación:** Disponer de una fecha completa en formato datetime permite utilizar las funcionalidades nativas de Pandas para análisis temporal (resampling, ventanas móviles, filtrado por rangos de fechas) sin necesidad de operaciones ad-hoc sobre las columnas individuales de año, mes y día.

```python
print("=" * 80)
print("3.2. CREACIÓN DE VARIABLE FECHA")
print("=" * 80)

has_year = df_prep['Año_int'] > 0
has_month = df_prep['Mes'].notna() & (df_prep['Mes'] > 0) & (df_prep['Mes'] <= 12)
has_day = df_prep['Día'].notna() & (df_prep['Día'] > 0) & (df_prep['Día'] <= 31)

mask_complete = has_year & has_month & has_day

df_prep['Fecha'] = pd.NaT

date_strings = (
    df_prep.loc[mask_complete, 'Año_int'].astype(str) + '-' +
    df_prep.loc[mask_complete, 'Mes'].astype(int).astype(str).str.zfill(2) + '-' +
    df_prep.loc[mask_complete, 'Día'].astype(int).astype(str).str.zfill(2)
)

df_prep.loc[mask_complete, 'Fecha'] = pd.to_datetime(date_strings, format='%Y-%m-%d', errors='coerce')

n_with_date = df_prep['Fecha'].notna().sum()
n_without_date = df_prep['Fecha'].isna().sum()

print(f"\n  Registros con fecha completa válida: {n_with_date:,} ({n_with_date/len(df_prep)*100:.1f}%)")
print(f"  Registros sin fecha completa:        {n_without_date:,} ({n_without_date/len(df_prep)*100:.1f}%)")

print("\n  Rango temporal del dataset:")
valid_dates = df_prep['Fecha'].dropna()
if len(valid_dates) > 0:
    print(f"    Fecha más antigua: {valid_dates.min().strftime('%Y-%m-%d')}")
    print(f"    Fecha más reciente: {valid_dates.max().strftime('%Y-%m-%d')}")

print(f"\n  Motivos de ausencia de fecha:")
print(f"    Año = 0 o nulo:    {(~has_year).sum():,} registros")
print(f"    Mes = 0 o nulo:    {(~has_month).sum():,} registros")
print(f"    Día = 0 o nulo:    {(~has_day).sum():,} registros")
```

### **3.3. Recodificación de categorías con baja representación**
---

Se revisan las variables categóricas principales para identificar categorías con frecuencia relativa inferior al 1% y agrupar en "OTROS" aquellas que no aporten valor analítico individual. Sin embargo, ciertas categorías de baja frecuencia se **excluyen explícitamente** de la agrupación por su relevancia histórica, política o militar en el contexto del conflicto armado colombiano.

**Criterio general:** Las categorías por debajo del umbral se agrupan en "OTROS", **salvo** aquellas cuya rareza misma es analíticamente significativa (eventos atípicos que merecen trazabilidad individual).

**Decisiones por variable:**

| Variable | Decisión | Categorías protegidas (no se agrupan en OTROS) | Justificación |
|----------|----------|------------------------------------------------|---------------|
| **Modalidad** | Agrupar baja frecuencia, con excepciones | `ATAQUE A POBLACIÓN` | En un dataset de acciones bélicas la población civil no es un objetivo militar legítimo; estos 11 registros tienen peso analítico especial para el estudio del conflicto y la memoria histórica. |
| **Grupo Armado 1** | Agrupar baja frecuencia, con excepciones | `GRUPO POSDESMOVILIZACIÓN`, `AGENTE EXTRANJERO`, `AGENTE DEL ESTADO - GRUPO PARAMILITAR` | Posdesmovilización es clave para el análisis del posacuerdo (AUC 2003-2006, FARC 2016). Agente extranjero documenta la dimensión internacional del conflicto interno. Los 3 registros Estado-paramilitar plantean interrogantes sobre posibles vínculos que merecen análisis individual. |
| **Grupo Armado 2** | Agrupar baja frecuencia, con excepciones | `AGENTE EXTRANJERO`, `AGENTE DEL ESTADO - GUERRILLA` | Agente extranjero (5 reg.) por las mismas razones. El único registro Estado-guerrilla es un caso excepcional que merece revisión individual. |
| **Descripción Grupo Armado 1** | Agrupar lista explícita | *(se agrupan solo: No identificada, Otro, Presencia GAO sin atribución, OTRA, OTRA DISIDENCIA, No aplica)* | Son categorías genéricas sin valor identificatorio. El resto de descripciones (frentes, bloques, brigadas específicas) se conservan íntegramente por su valor histórico. |
| **Descripción Grupo Armado 2** | Agrupar lista explícita | *(se agrupan solo: OTRO, OTRA, Presencia GAO sin atribución, Otra disidencia, No aplica)* | Misma justificación: categorías genéricas que no identifican un actor concreto. |
| **Iniciativa / Ventaja Militar** | Sin cambios | — | Todas las categorías válidas superan el umbral. |

```python
print("=" * 80)
print("3.3. RECODIFICACIÓN DE CATEGORÍAS — ANÁLISIS POR VARIABLE")
print("=" * 80)

threshold_pct = 1.0
threshold_abs = int(len(df_prep) * threshold_pct / 100)
print(f"\nUmbral de baja representación: {threshold_pct}% ({threshold_abs:,} registros)")

def recode_low_freq(df, col, threshold, protect=None):
    """Agrupa categorías por debajo del umbral en 'OTROS', protegiendo las indicadas."""
    if col not in df.columns:
        return 0
    protect = set(protect or [])
    counts = df[col].value_counts(dropna=True)
    low_freq = counts[counts < threshold]
    to_recode = [cat for cat in low_freq.index if cat not in protect]

    if not to_recode:
        return 0

    if hasattr(df[col], 'cat'):
        df[col] = df[col].cat.add_categories('OTROS')

    mask = df[col].isin(to_recode)
    n = mask.sum()
    df.loc[mask, col] = 'OTROS'

    if hasattr(df[col], 'cat'):
        df[col] = df[col].cat.remove_unused_categories()

    return n

# ═══════════════════════════════════════════════════════════════════════════
# MODALIDAD
# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("MODALIDAD")
print("=" * 80)

modalidad_protect = ['ATAQUE A POBLACIÓN']
print(f"\n  Categorías protegidas (no se agrupan en OTROS):")
for p in modalidad_protect:
    n = (df_prep['Modalidad'] == p).sum()
    print(f"    ✦ '{p}': {n} registros — La población civil no es un objetivo")
    print(f"      militar legítimo; estos casos tienen peso analítico especial")
    print(f"      para el estudio del conflicto y la memoria histórica.")

counts_mod = df_prep['Modalidad'].value_counts(dropna=True)
low_mod = counts_mod[counts_mod < threshold_abs]
to_group_mod = [c for c in low_mod.index if c not in modalidad_protect]
print(f"\n  Categorías agrupadas en OTROS:")
for c in to_group_mod:
    print(f"    → '{c}': {low_mod[c]} registros")

n_mod = recode_low_freq(df_prep, 'Modalidad', threshold_abs, protect=modalidad_protect)
print(f"\n  Registros reclasificados: {n_mod:,}")
print(f"  Categorías únicas después: {df_prep['Modalidad'].nunique()}")

# ═══════════════════════════════════════════════════════════════════════════
# GRUPO ARMADO 1
# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("GRUPO ARMADO 1")
print("=" * 80)

ga1_protect = ['GRUPO POSDESMOVILIZACIÓN', 'AGENTE EXTRANJERO',
               'AGENTE DEL ESTADO - GRUPO PARAMILITAR']
print(f"\n  Categorías protegidas (no se agrupan en OTROS):")
for p in ga1_protect:
    n = (df_prep['Grupo Armado 1'] == p).sum()
    print(f"    ✦ '{p}': {n} registros")

if 'GRUPO POSDESMOVILIZACIÓN' in ga1_protect:
    print(f"      → Clave para el análisis del posacuerdo: qué grupos retomaron")
    print(f"        dinámicas de enfrentamiento tras la desmovilización (AUC 2003-2006, FARC 2016).")
if 'AGENTE EXTRANJERO' in ga1_protect:
    print(f"      → La participación de agentes extranjeros en el conflicto armado")
    print(f"        interno colombiano es excepcional y requiere análisis de su origen y acción.")
if 'AGENTE DEL ESTADO - GRUPO PARAMILITAR' in ga1_protect:
    print(f"      → Estos registros donde un actor se clasifica simultáneamente como")
    print(f"        Estado y paramilitar merecen análisis por las implicaciones sobre posibles vínculos.")

counts_ga1 = df_prep['Grupo Armado 1'].value_counts(dropna=True)
low_ga1 = counts_ga1[counts_ga1 < threshold_abs]
to_group_ga1 = [c for c in low_ga1.index if c not in ga1_protect]
print(f"\n  Categorías agrupadas en OTROS:")
for c in to_group_ga1:
    print(f"    → '{c}': {low_ga1[c]} registros")

n_ga1 = recode_low_freq(df_prep, 'Grupo Armado 1', threshold_abs, protect=ga1_protect)
print(f"\n  Registros reclasificados: {n_ga1:,}")
print(f"  Categorías únicas después: {df_prep['Grupo Armado 1'].nunique()}")

# ═══════════════════════════════════════════════════════════════════════════
# GRUPO ARMADO 2
# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("GRUPO ARMADO 2")
print("=" * 80)

ga2_protect = ['AGENTE EXTRANJERO', 'AGENTE DEL ESTADO - GUERRILLA']
print(f"\n  Categorías protegidas (no se agrupan en OTROS):")
for p in ga2_protect:
    n = (df_prep['Grupo Armado 2'] == p).sum()
    print(f"    ✦ '{p}': {n} registros")

if 'AGENTE EXTRANJERO' in ga2_protect:
    print(f"      → Misma justificación que en Grupo Armado 1: excepcionalidad de")
    print(f"        la participación de un agente extranjero en el conflicto interno.")
if 'AGENTE DEL ESTADO - GUERRILLA' in ga2_protect:
    print(f"      → Registro excepcional donde un actor aparece como Estado y guerrilla")
    print(f"        simultáneamente; requiere revisión individual por su carácter único.")

counts_ga2 = df_prep['Grupo Armado 2'].value_counts(dropna=True)
low_ga2 = counts_ga2[counts_ga2 < threshold_abs]
to_group_ga2 = [c for c in low_ga2.index if c not in ga2_protect]
print(f"\n  Categorías agrupadas en OTROS:")
for c in to_group_ga2:
    print(f"    → '{c}': {low_ga2[c]} registros")

n_ga2 = recode_low_freq(df_prep, 'Grupo Armado 2', threshold_abs, protect=ga2_protect)
print(f"\n  Registros reclasificados: {n_ga2:,}")
print(f"  Categorías únicas después: {df_prep['Grupo Armado 2'].nunique()}")

# ═══════════════════════════════════════════════════════════════════════════
# DESCRIPCIÓN GRUPO ARMADO 1 — lista explícita
# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("DESCRIPCIÓN GRUPO ARMADO 1 — Agrupación por lista explícita")
print("=" * 80)

desc_ga1_to_otros = ['NO IDENTIFICADA', 'OTRO', 'PRESENCIA GAO SIN ATRIBUCIÓN',
                     'OTRA', 'OTRA DISIDENCIA', 'NO APLICA']

if 'Descripción Grupo Armado 1' in df_prep.columns:
    if hasattr(df_prep['Descripción Grupo Armado 1'], 'cat'):
        df_prep['Descripción Grupo Armado 1'] = df_prep['Descripción Grupo Armado 1'].cat.add_categories('OTROS')

    mask_dga1 = df_prep['Descripción Grupo Armado 1'].isin(desc_ga1_to_otros)
    n_dga1 = mask_dga1.sum()
    print(f"\n  Categorías genéricas agrupadas en 'OTROS': {desc_ga1_to_otros}")
    for c in desc_ga1_to_otros:
        n = (df_prep['Descripción Grupo Armado 1'] == c).sum()
        if n > 0:
            print(f"    → '{c}': {n} registros")
    print(f"  Total registros afectados: {n_dga1:,}")
    df_prep.loc[mask_dga1, 'Descripción Grupo Armado 1'] = 'OTROS'

    if hasattr(df_prep['Descripción Grupo Armado 1'], 'cat'):
        df_prep['Descripción Grupo Armado 1'] = df_prep['Descripción Grupo Armado 1'].cat.remove_unused_categories()

    print(f"  Categorías únicas después: {df_prep['Descripción Grupo Armado 1'].nunique()}")
    print(f"  Todas las demás descripciones (frentes, bloques, brigadas) se conservan")
    print(f"  individualmente por su valor histórico.")

# ═══════════════════════════════════════════════════════════════════════════
# DESCRIPCIÓN GRUPO ARMADO 2 — lista explícita
# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("DESCRIPCIÓN GRUPO ARMADO 2 — Agrupación por lista explícita")
print("=" * 80)

desc_ga2_to_otros = ['OTRO', 'OTRA', 'PRESENCIA GAO SIN ATRIBUCIÓN',
                     'OTRA DISIDENCIA', 'NO APLICA']

if 'Descripción Grupo Armado 2' in df_prep.columns:
    if hasattr(df_prep['Descripción Grupo Armado 2'], 'cat'):
        df_prep['Descripción Grupo Armado 2'] = df_prep['Descripción Grupo Armado 2'].cat.add_categories('OTROS')

    mask_dga2 = df_prep['Descripción Grupo Armado 2'].isin(desc_ga2_to_otros)
    n_dga2 = mask_dga2.sum()
    print(f"\n  Categorías genéricas agrupadas en 'OTROS': {desc_ga2_to_otros}")
    for c in desc_ga2_to_otros:
        n = (df_prep['Descripción Grupo Armado 2'] == c).sum()
        if n > 0:
            print(f"    → '{c}': {n} registros")
    print(f"  Total registros afectados: {n_dga2:,}")
    df_prep.loc[mask_dga2, 'Descripción Grupo Armado 2'] = 'OTROS'

    if hasattr(df_prep['Descripción Grupo Armado 2'], 'cat'):
        df_prep['Descripción Grupo Armado 2'] = df_prep['Descripción Grupo Armado 2'].cat.remove_unused_categories()

    print(f"  Categorías únicas después: {df_prep['Descripción Grupo Armado 2'].nunique()}")
    print(f"  Todas las demás descripciones se conservan individualmente.")

# ═══════════════════════════════════════════════════════════════════════════
print(f"\n{'=' * 80}")
print("INICIATIVA / VENTAJA MILITAR: Sin cambios necesarios — todas las categorías")
print("válidas superan el umbral.")
print("=" * 80)
```

```python
from scipy import stats

num_vars = ['Año_int', 'Total de Víctimas del Caso', 'Lesionados Civiles',
            'Lesionados Combatientes', 'Capturados',
            'Muertes Total Civiles', 'Muertes Total Combatientes',
            'Muertes Total Agentes del Estado', 'Muertes Total Combatientes GAO',
            'Tiene_Víctimas']

df_num = df_prep[num_vars].dropna()

print("=" * 80)
print("1.1.1. MATRIZ DE CORRELACIÓN DE SPEARMAN — VARIABLES NUMÉRICAS")
print("=" * 80)

spearman_corr = df_num.corr(method='spearman')

fig, ax = plt.subplots(figsize=(14, 10))
mask = np.triu(np.ones_like(spearman_corr, dtype=bool), k=1)
sns.heatmap(spearman_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, square=True, linewidths=0.5,
            cbar_kws={'label': 'Coeficiente de Spearman'}, ax=ax)
ax.set_title('Matriz de correlación de Spearman — Variables numéricas del conflicto', fontsize=14, pad=20)
plt.tight_layout()
save_fig(fig, 'fase4_spearman_correlation')
plt.show()

print("\nCorrelaciones más fuertes (|ρ| > 0.5, excluyendo diagonal):")
for i in range(len(spearman_corr)):
    for j in range(i+1, len(spearman_corr)):
        val = spearman_corr.iloc[i, j]
        if abs(val) > 0.5:
            print(f"  {spearman_corr.index[i]} ↔ {spearman_corr.columns[j]}: ρ = {val:.3f}")
```

#### **1.1.2. Correlaciones entre variables categóricas (Chi-cuadrado y Cramér's V)**

Las variables categóricas del conflicto (Modalidad, Grupo Armado, Iniciativa, Departamento, etc.) no pueden analizarse con correlación de Pearson/Spearman. Se utiliza el **test Chi-cuadrado de independencia** para evaluar si existe asociación estadísticamente significativa entre pares de variables categóricas, y el **coeficiente V de Cramér** como medida del tamaño del efecto (0 = sin asociación, 1 = asociación perfecta).

**Justificación:** Este análisis permite identificar qué factores del conflicto están estadísticamente relacionados entre sí, lo cual es fundamental para entender las dinámicas del conflicto y para seleccionar variables predictoras en los modelos de regresión.

```python
from scipy.stats import chi2_contingency

def cramers_v(var1, var2):
    confusion_matrix = pd.crosstab(var1, var2)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    min_dim = min(confusion_matrix.shape) - 1
    if min_dim == 0 or n == 0:
        return 0.0
    return np.sqrt(chi2 / (n * min_dim))

cat_vars_analysis = ['Modalidad', 'Grupo Armado 1', 'Grupo Armado 2',
                     'Iniciativa', 'Ventaja Militar', 'Departamento']

df_cat_clean = df_prep[cat_vars_analysis].dropna()

print("=" * 80)
print("1.1.2. ASOCIACIÓN ENTRE VARIABLES CATEGÓRICAS (CHI² y CRAMÉR'S V)")
print("=" * 80)

n_vars = len(cat_vars_analysis)
cramer_matrix = pd.DataFrame(np.zeros((n_vars, n_vars)),
                              index=cat_vars_analysis, columns=cat_vars_analysis)
pvalue_matrix = pd.DataFrame(np.ones((n_vars, n_vars)),
                              index=cat_vars_analysis, columns=cat_vars_analysis)

for i in range(n_vars):
    for j in range(i, n_vars):
        if i == j:
            cramer_matrix.iloc[i, j] = 1.0
            pvalue_matrix.iloc[i, j] = 0.0
        else:
            v1 = df_cat_clean[cat_vars_analysis[i]]
            v2 = df_cat_clean[cat_vars_analysis[j]]
            ct = pd.crosstab(v1, v2)
            chi2, p, dof, expected = chi2_contingency(ct)
            cv = cramers_v(v1, v2)
            cramer_matrix.iloc[i, j] = cv
            cramer_matrix.iloc[j, i] = cv
            pvalue_matrix.iloc[i, j] = p
            pvalue_matrix.iloc[j, i] = p

fig, ax = plt.subplots(figsize=(10, 8))
mask = np.triu(np.ones_like(cramer_matrix, dtype=bool), k=1)
sns.heatmap(cramer_matrix.astype(float), mask=mask, annot=True, fmt='.3f', cmap='YlOrRd',
            vmin=0, vmax=1, square=True, linewidths=0.5,
            cbar_kws={'label': "V de Cramér"}, ax=ax)
ax.set_title("Asociación entre variables categóricas (V de Cramér)", fontsize=14, pad=20)
plt.tight_layout()
save_fig(fig, 'fase4_cramers_v_matrix')
plt.show()

alpha = 0.05
n_tests = n_vars * (n_vars - 1) // 2
alpha_bonferroni = alpha / n_tests
print(f"\nCorrección de Bonferroni: α ajustado = {alpha}/{n_tests} = {alpha_bonferroni:.6f}")

print(f"\nAsociaciones significativas (p < {alpha_bonferroni:.6f}):")
for i in range(n_vars):
    for j in range(i+1, n_vars):
        p = pvalue_matrix.iloc[i, j]
        cv = cramer_matrix.iloc[i, j]
        sig = "✓" if p < alpha_bonferroni else "✗"
        if p < alpha_bonferroni:
            strength = "fuerte" if cv > 0.3 else "moderada" if cv > 0.1 else "débil"
            print(f"  {sig} {cat_vars_analysis[i]} ↔ {cat_vars_analysis[j]}: "
                  f"V = {cv:.3f} ({strength}), p = {p:.2e}")
```

#### **1.1.3. Asociación entre variables categóricas y la variable objetivo `Tiene_Víctimas`**

Se analiza específicamente la asociación entre cada variable categórica del contexto del evento y la variable binaria `Tiene_Víctimas`, para identificar qué factores están más relacionados con la ocurrencia de víctimas. Esto informa directamente la selección de predictores para el modelo de regresión logística.

```python
print("=" * 80)
print("1.1.3. ASOCIACIÓN DE VARIABLES CATEGÓRICAS CON Tiene_Víctimas")
print("=" * 80)

target = 'Tiene_Víctimas'
cat_predictors = ['Modalidad', 'Grupo Armado 1', 'Grupo Armado 2',
                  'Iniciativa', 'Ventaja Militar', 'Departamento', 'Década']

results_cat = []
for col in cat_predictors:
    if col not in df_prep.columns:
        continue
    sub = df_prep[[col, target]].dropna()
    ct = pd.crosstab(sub[col], sub[target])
    if ct.shape[0] < 2 or ct.shape[1] < 2:
        continue
    chi2, p, dof, expected = chi2_contingency(ct)
    cv = cramers_v(sub[col], sub[target])
    results_cat.append({'Variable': col, 'Chi²': chi2, 'p-valor': p,
                        'DoF': dof, "V de Cramér": cv})

results_cat_df = pd.DataFrame(results_cat).sort_values("V de Cramér", ascending=False)
results_cat_df['Significativa'] = results_cat_df['p-valor'] < alpha_bonferroni
display(results_cat_df.reset_index(drop=True))

fig, ax = plt.subplots(figsize=(10, 5))
colors = ['#2ecc71' if s else '#e74c3c' for s in results_cat_df['Significativa']]
bars = ax.barh(results_cat_df['Variable'], results_cat_df["V de Cramér"], color=colors)
ax.set_xlabel("V de Cramér")
ax.set_title(f"Fuerza de asociación con '{target}' (verde = significativa, rojo = no)", fontsize=13)
ax.axvline(x=0.1, color='gray', linestyle='--', alpha=0.5, label='Umbral débil (0.1)')
ax.axvline(x=0.3, color='gray', linestyle=':', alpha=0.5, label='Umbral moderado (0.3)')
ax.legend()
plt.tight_layout()
save_fig(fig, 'fase4_cramers_v_vs_target')
plt.show()

print("\nInterpretación:")
for _, row in results_cat_df.iterrows():
    strength = "fuerte" if row["V de Cramér"] > 0.3 else "moderada" if row["V de Cramér"] > 0.1 else "débil"
    sig = "significativa" if row['Significativa'] else "no significativa"
    print(f"  {row['Variable']}: asociación {strength} ({sig}), V = {row['V de Cramér']:.3f}")
```

### **1.2. Análisis de regresiones**
---

A partir de las correlaciones identificadas, se plantean modelos de regresión para estimar variables de interés. Dado que la variable objetivo principal es **`Tiene_Víctimas`** (binaria: 0/1), se utiliza **regresión logística**, que es la técnica apropiada para modelar la probabilidad de un evento dicotómico.

**Selección de variables predictoras:** Se seleccionan las variables que mostraron asociación significativa con `Tiene_Víctimas` en la sección anterior, junto con la variable temporal `Año_int` que permite capturar la evolución del conflicto.

**Justificación del modelo:** La regresión logística permite cuantificar cómo las características del evento (modalidad, actores, época, ubicación) influyen en la probabilidad de que se produzcan víctimas, lo cual es directamente relevante para los objetivos de comprensión del conflicto definidos en la [Fase 1](#fase1).

```python
import statsmodels.api as sm
from sklearn.preprocessing import LabelEncoder

print("=" * 80)
print("1.2. REGRESIÓN LOGÍSTICA — Tiene_Víctimas")
print("=" * 80)

target_col = 'Tiene_Víctimas'
cat_features_reg = ['Modalidad', 'Grupo Armado 1', 'Iniciativa']
num_features_reg = ['Año_int']

df_reg = df_prep[cat_features_reg + num_features_reg + [target_col]].dropna().copy()
print(f"\nRegistros para el modelo (tras eliminar nulos): {len(df_reg):,}")

print(f"\nDistribución de la variable objetivo:")
vc = df_reg[target_col].value_counts()
for val, count in vc.items():
    print(f"  {target_col} = {val}: {count:,} ({count/len(df_reg)*100:.1f}%)")

df_reg = df_reg.reset_index(drop=True)
df_dummies = pd.get_dummies(df_reg[cat_features_reg], drop_first=True, dtype=int)
X = pd.concat([df_dummies, df_reg[num_features_reg]], axis=1)
X = sm.add_constant(X)
y = df_reg[target_col]

print(f"\nVariables predictoras en el modelo: {X.shape[1]} (incluyendo constante)")
print(f"  Categóricas codificadas (one-hot): {df_dummies.shape[1]}")
print(f"  Numéricas: {len(num_features_reg)}")

logit_model = sm.Logit(y, X)
logit_result = logit_model.fit(disp=0, maxiter=100)

print(f"\n{'=' * 80}")
print("RESUMEN DEL MODELO DE REGRESIÓN LOGÍSTICA")
print(f"{'=' * 80}")
print(logit_result.summary2())
```

```python
print("=" * 80)
print("EVALUACIÓN DEL MODELO")
print("=" * 80)

print(f"\nPseudo R² (McFadden): {logit_result.prsquared:.4f}")
print(f"Log-Likelihood: {logit_result.llf:.2f}")
print(f"AIC: {logit_result.aic:.2f}")
print(f"BIC: {logit_result.bic:.2f}")

y_pred_prob = logit_result.predict(X)
y_pred = (y_pred_prob >= 0.5).astype(int)
accuracy = (y_pred == y).mean()
print(f"Accuracy (umbral 0.5): {accuracy:.4f} ({accuracy*100:.1f}%)")

from sklearn.metrics import classification_report, confusion_matrix

cm = confusion_matrix(y, y_pred)
print(f"\nMatriz de confusión:")
print(f"  {'':>20} Predicho=0  Predicho=1")
print(f"  {'Real=0':<20} {cm[0,0]:>10,}  {cm[0,1]:>10,}")
print(f"  {'Real=1':<20} {cm[1,0]:>10,}  {cm[1,1]:>10,}")

print(f"\nReporte de clasificación:")
print(classification_report(y, y_pred, target_names=['Sin víctimas (0)', 'Con víctimas (1)']))

sig_vars = logit_result.pvalues[logit_result.pvalues < 0.05].index.tolist()
print(f"\nVariables significativas (p < 0.05): {len(sig_vars)}")
for var in sig_vars:
    coef = logit_result.params[var]
    p = logit_result.pvalues[var]
    odds_ratio = np.exp(coef)
    direction = "aumenta" if coef > 0 else "disminuye"
    print(f"  {var}: coef = {coef:.4f}, OR = {odds_ratio:.4f}, p = {p:.4e} "
          f"→ {direction} la probabilidad de víctimas")
```

### **1.3. Estadística inferencial**
---

Se realizan pruebas de hipótesis para contrastar formalmente patrones observados en el análisis exploratorio de la [Fase 2](#fase2). Se evalúa si las diferencias observadas en la distribución de víctimas entre grupos de actores, modalidades y periodos temporales son estadísticamente significativas o podrían deberse al azar.

#### **1.3.1. Test de Wald sobre los coeficientes del modelo logístico**

El test de Wald (ya incluido en el resumen del modelo) evalúa si cada coeficiente es significativamente diferente de cero. Los coeficientes significativos indican variables que tienen un efecto real sobre la probabilidad de víctimas.

#### **1.3.2. Prueba de Kruskal-Wallis: Víctimas por Modalidad**

Se utiliza Kruskal-Wallis (alternativa no paramétrica al ANOVA) dado que las distribuciones de víctimas son fuertemente asimétricas (como se documentó en la [Fase 2, sección 2.2](#fase2)).

```python
from scipy.stats import kruskal, mannwhitneyu

print("=" * 80)
print("1.3.2. PRUEBA DE KRUSKAL-WALLIS: VÍCTIMAS POR MODALIDAD")
print("=" * 80)

groups_mod = []
labels_mod = []
for mod, group in df_prep.groupby('Modalidad')['Total de Víctimas del Caso']:
    data = group.dropna().values
    if len(data) >= 5:
        groups_mod.append(data)
        labels_mod.append(mod)

stat_kw, p_kw = kruskal(*groups_mod)
print(f"\nH₀: La distribución de víctimas es igual entre todas las modalidades")
print(f"H₁: Al menos una modalidad tiene una distribución de víctimas diferente")
print(f"\nEstadístico H: {stat_kw:.2f}")
print(f"p-valor: {p_kw:.2e}")
print(f"Decisión: {'Se rechaza H₀' if p_kw < 0.05 else 'No se rechaza H₀'} (α = 0.05)")

if p_kw < 0.05:
    print(f"\n→ Existen diferencias estadísticamente significativas en la distribución")
    print(f"  de víctimas entre las distintas modalidades de acción bélica.")

    print(f"\n  Mediana de víctimas por modalidad:")
    medians = df_prep.groupby('Modalidad')['Total de Víctimas del Caso'].median().sort_values(ascending=False)
    for mod, med in medians.items():
        n = (df_prep['Modalidad'] == mod).sum()
        print(f"    {mod:<55} mediana = {med:.0f}  (n = {n:,})")

print(f"\n{'=' * 80}")
print("PRUEBA DE KRUSKAL-WALLIS: VÍCTIMAS POR GRUPO ARMADO 1")
print("=" * 80)

groups_ga1 = []
labels_ga1 = []
for ga, group in df_prep.groupby('Grupo Armado 1')['Total de Víctimas del Caso']:
    data = group.dropna().values
    if len(data) >= 5:
        groups_ga1.append(data)
        labels_ga1.append(ga)

stat_ga1, p_ga1 = kruskal(*groups_ga1)
print(f"\nH₀: La distribución de víctimas es igual entre todos los grupos armados")
print(f"H₁: Al menos un grupo armado tiene una distribución de víctimas diferente")
print(f"\nEstadístico H: {stat_ga1:.2f}")
print(f"p-valor: {p_ga1:.2e}")
print(f"Decisión: {'Se rechaza H₀' if p_ga1 < 0.05 else 'No se rechaza H₀'} (α = 0.05)")

print(f"\n{'=' * 80}")
print("PRUEBA DE MANN-WHITNEY U: VÍCTIMAS ANTES vs DESPUÉS DEL ACUERDO DE PAZ (2016)")
print("=" * 80)

df_valid_year = df_prep[df_prep['Año_int'] > 0].copy()
before_peace = df_valid_year[df_valid_year['Año_int'] < 2016]['Total de Víctimas del Caso'].dropna()
after_peace = df_valid_year[df_valid_year['Año_int'] >= 2016]['Total de Víctimas del Caso'].dropna()

stat_mw, p_mw = mannwhitneyu(before_peace, after_peace, alternative='two-sided')
print(f"\nH₀: La distribución de víctimas es igual antes y después de 2016")
print(f"H₁: Las distribuciones son diferentes")
print(f"\nAntes de 2016: n = {len(before_peace):,}, mediana = {before_peace.median():.0f}, media = {before_peace.mean():.2f}")
print(f"Desde 2016:    n = {len(after_peace):,}, mediana = {after_peace.median():.0f}, media = {after_peace.mean():.2f}")
print(f"\nEstadístico U: {stat_mw:.2f}")
print(f"p-valor: {p_mw:.2e}")
print(f"Decisión: {'Se rechaza H₀' if p_mw < 0.05 else 'No se rechaza H₀'} (α = 0.05)")
```

A diferencia del mapa de puntos de la Fase 2 —que muestra la ubicación individual de cada evento—, el coroplético sintetiza patrones geográficos a nivel departamental, facilitando la identificación de zonas críticas para la focalización de políticas públicas.

```python
import json as _json

GEOJSON_URL = "https://raw.githubusercontent.com/santiblanko/colombia.geojson/master/depto.json"
geojson_path = IMAGES_DIR / 'colombia_deptos.geojson'

if not geojson_path.exists():
    print("Descargando GeoJSON de departamentos de Colombia...")
    import urllib.request
    urllib.request.urlretrieve(GEOJSON_URL, geojson_path)

with open(geojson_path, 'r') as f:
    colombia_geo = _json.load(f)

dept_name_map = {}
for feat in colombia_geo['features']:
    props = feat['properties']
    name_key = None
    for k in ['NOMBRE_DPT', 'DEPARTAMEN', 'name', 'DPTO']:
        if k in props:
            name_key = k
            break
    if name_key:
        dept_name_map[props[name_key].upper()] = props[name_key]
        feat['id'] = props[name_key].upper()
    else:
        feat['id'] = str(feat.get('id', ''))

print(f"GeoJSON cargado: {len(colombia_geo['features'])} departamentos")
print(f"Nombres disponibles: {sorted(dept_name_map.keys())[:10]}...")

df_dept = df_prep[df_prep['Departamento'].notna() & (df_prep['Departamento'] != 'EXTERIOR')].copy()

dept_stats = df_dept.groupby('Departamento').agg(
    n_eventos=('Tiene_Víctimas', 'count'),
    tasa_victimas=('Tiene_Víctimas', 'mean'),
    total_victimas=('Total de Víctimas del Caso', 'sum'),
    media_victimas=('Total de Víctimas del Caso', 'mean'),
    n_modalidades=('Modalidad', 'nunique'),
).reset_index()

dept_stats['Departamento_upper'] = dept_stats['Departamento'].str.upper().str.strip()

print(f"\nDepartamentos con datos: {len(dept_stats)}")
print(f"Total eventos incluidos: {dept_stats['n_eventos'].sum():,}")
```

```python
fig_choro1 = px.choropleth(
    dept_stats,
    geojson=colombia_geo,
    locations='Departamento_upper',
    featureidkey='id',
    color='n_eventos',
    color_continuous_scale='YlOrRd',
    scope='south america',
    hover_name='Departamento',
    hover_data={
        'Departamento_upper': False,
        'n_eventos': ':,',
        'tasa_victimas': ':.1%',
        'total_victimas': ':,',
        'media_victimas': ':.2f',
    },
    labels={
        'n_eventos': 'Total eventos',
        'tasa_victimas': 'Tasa con víctimas',
        'total_victimas': 'Total víctimas',
        'media_victimas': 'Media víctimas/evento',
    },
    title='Intensidad del conflicto armado por departamento — Número de acciones bélicas',
)
fig_choro1.update_geos(
    fitbounds='locations',
    visible=False,
    bgcolor='rgba(0,0,0,0)',
)
fig_choro1.update_layout(
    width=700, height=800,
    margin=dict(l=0, r=0, t=50, b=0),
    coloraxis_colorbar=dict(title='N° eventos', len=0.5),
)
fig_choro1.show()

fig_choro2 = px.choropleth(
    dept_stats,
    geojson=colombia_geo,
    locations='Departamento_upper',
    featureidkey='id',
    color='tasa_victimas',
    color_continuous_scale='RdYlGn_r',
    scope='south america',
    hover_name='Departamento',
    hover_data={
        'Departamento_upper': False,
        'n_eventos': ':,',
        'tasa_victimas': ':.1%',
        'total_victimas': ':,',
    },
    labels={
        'n_eventos': 'Total eventos',
        'tasa_victimas': 'Tasa con víctimas',
        'total_victimas': 'Total víctimas',
    },
    title='Tasa de eventos con víctimas por departamento',
)
fig_choro2.update_geos(fitbounds='locations', visible=False, bgcolor='rgba(0,0,0,0)')
fig_choro2.update_layout(
    width=700, height=800,
    margin=dict(l=0, r=0, t=50, b=0),
    coloraxis_colorbar=dict(title='Tasa víctimas', len=0.5, tickformat='.0%'),
)
fig_choro2.show()
```

#### **Mapa coroplético comparativo: antes y después del Acuerdo de Paz (2016)**

Se genera un mapa interactivo que permite comparar la **intensidad anualizada** del conflicto por departamento en dos periodos: antes de 2016 y desde 2016. Se utiliza la métrica **eventos por año** (total de eventos dividido entre el número de años del periodo) en lugar de la suma absoluta, ya que los periodos tienen duraciones muy distintas (~58 años vs. ~8 años) y una comparación de totales sería conceptualmente incorrecta.

Esto complementa el hallazgo estadístico de la prueba de Mann-Whitney (sección 1.3) con una dimensión geográfica: no solo hubo una reducción general, sino que se puede identificar en qué departamentos fue más pronunciada.

```python
df_valid_yr = df_dept[df_dept['Año_int'] > 0].copy()

n_years_before = df_valid_yr[df_valid_yr['Año_int'] < 2016]['Año_int'].nunique()
n_years_after = df_valid_yr[df_valid_yr['Año_int'] >= 2016]['Año_int'].nunique()
print(f"Años distintos antes de 2016: {n_years_before}")
print(f"Años distintos desde 2016:    {n_years_after}")

def build_period_stats(df_src, periodo_label, n_years):
    agg = df_src.groupby('Departamento').agg(
        n_eventos=('Tiene_Víctimas', 'count'),
        tasa_victimas=('Tiene_Víctimas', 'mean'),
    ).reset_index()
    agg['eventos_por_año'] = (agg['n_eventos'] / n_years).round(1)
    agg['Departamento_upper'] = agg['Departamento'].str.upper().str.strip()
    agg['periodo'] = periodo_label
    return agg

df_before = build_period_stats(
    df_valid_yr[df_valid_yr['Año_int'] < 2016], 'Antes de 2016', n_years_before)
df_after = build_period_stats(
    df_valid_yr[df_valid_yr['Año_int'] >= 2016], 'Desde 2016', n_years_after)

df_periods = pd.concat([df_before, df_after], ignore_index=True)

fig_rate = px.choropleth(
    df_periods,
    geojson=colombia_geo,
    locations='Departamento_upper',
    featureidkey='id',
    color='eventos_por_año',
    animation_frame='periodo',
    color_continuous_scale='YlOrRd',
    scope='south america',
    hover_name='Departamento',
    hover_data={
        'Departamento_upper': False,
        'eventos_por_año': ':.1f',
        'n_eventos': ':,',
        'tasa_victimas': ':.1%',
    },
    labels={
        'eventos_por_año': 'Eventos/año',
        'n_eventos': 'Total eventos',
        'tasa_victimas': 'Tasa con víctimas',
    },
    title='Intensidad anualizada del conflicto por departamento: antes vs. después del Acuerdo de Paz',
    range_color=[0, df_periods['eventos_por_año'].quantile(0.95)],
)
fig_rate.update_geos(fitbounds='locations', visible=False, bgcolor='rgba(0,0,0,0)')
fig_rate.update_layout(
    width=750, height=850,
    margin=dict(l=0, r=0, t=60, b=0),
    coloraxis_colorbar=dict(title='Eventos/año', len=0.5),
)
fig_rate.show()

print("=" * 80)
print("RESUMEN COMPARATIVO — TASA ANUALIZADA (eventos/año)")
print("=" * 80)
for periodo, grupo in df_periods.groupby('periodo'):
    n_yrs = n_years_before if 'Antes' in periodo else n_years_after
    top5 = grupo.nlargest(5, 'eventos_por_año')[['Departamento', 'n_eventos', 'eventos_por_año', 'tasa_victimas']]
    print(f"\n  {periodo} ({n_yrs} años) — Top 5 departamentos:")
    for _, row in top5.iterrows():
        print(f"    {row['Departamento']:<25} {row['eventos_por_año']:>6.1f} eventos/año "
              f"(total: {row['n_eventos']:>5,}, tasa víctimas: {row['tasa_victimas']:.1%})")
```

#### Asociación entre variables categóricas

- **Modalidad ↔ Iniciativa** (V de Cramér fuerte): Los agentes del Estado y los grupos armados organizados tienden a emplear modalidades distintas cuando toman la iniciativa del enfrentamiento. Esto indica que cada tipo de actor privilegia modalidades específicas según las armas a su disposición, su capacidad operativa y su agencia táctica. Este hallazgo permite comprender los perfiles de combate diferenciados de cada actor.

- **Modalidad ↔ Ventaja Militar** (V de Cramér fuerte): La modalidad del enfrentamiento está estrechamente relacionada con quién obtiene la ventaja militar. Ciertas modalidades (como la emboscada o el bombardeo aéreo) confieren ventaja al actor que las ejecuta, lo que permite identificar qué tácticas están asociadas a una posición dominante en el enfrentamiento.

- **Ventaja Militar ↔ Iniciativa** (V de Cramér ≈ 1.0): Se observa una correlación prácticamente perfecta: quien tiene la iniciativa del enfrentamiento obtiene sistemáticamente la ventaja militar. Esto sugiere que la preparación y planificación previa que implica tomar la iniciativa otorga una ventaja decisiva sobre el actor que es sorprendido o que reacciona al ataque.

#### Fuerza de asociación con `Tiene_Víctimas`
